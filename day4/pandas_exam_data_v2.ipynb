```python
import pandas as pd
import numpy as np
df=pd.read_csv('pandas_exam_data_v2.csv')
df
```

```python
df=df.fillna({"工资": df["工资"].median()})
df
```

```python
df = df.drop_duplicates(subset=["姓名"]) 
df
```

```python
df=df.drop(columns=['姓名'])
df
```

```python
df['工龄']=df['工龄'].astype(float)
df
```

```python
dsum=df['工资'].sum()
mean=df['工资'].mean()
std = df['工资'].std()

print(dsum,
mean,
std)
```

```python
opgroup= df.groupby("部门").agg(平均工资=('工资','mean') ,平均年龄=('年龄','mean'))
opgroup
```

```python
df['年龄段']=pd.cut(df['年龄'],bins=[0,30,40,60],labels=["青年", "中年", "中老年"])
df
```

```python
df.value_counts("部门")
```

```python
df['薪级别']=np.where(df['工资'] > 8000,"高新","普通")
df
```
