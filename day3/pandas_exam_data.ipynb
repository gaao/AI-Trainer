```python
import pandas as pd
import numpy as np
# df=pd.read_csv('pandas_exam_data.csv').head()
df=pd.read_csv('pandas_exam_data.csv')
```

```python
pd.isnull(df)
df.dropna()
```

```python
df.fillna({"工资": df["工资"].mean()})
```

```python
# df['是否重复'] = df.duplicated(keep=False)
df = df.drop_duplicates(subset=["姓名"])
df
```

```python
df = df.fillna(value={"年龄": 23})
df["年龄"] = df["年龄"].astype(int)
df
```

```python
gzsum= df["工资"].sum()
gzmean= df["工资"].mean()
gzstd= df["工资"].std()
print(gzsum,gzmean,gzstd)
```

```python
df.groupby("部门").agg(平均工资=("工资", "mean"),总人数=("姓名", "count"))
```

```python
df["年龄段"] = pd.cut(
    df["年龄"],
    bins=[0, 25, 35, 50],
    labels=["青年", "中年", "中老年"]
)
df
```

```python
df["部门"].value_counts()
```

```python
df['收入登记']=np.where(df['工资']>=7000,'高收入','一般')
df
```

```python
bt=df[df['工资'].between(6000, 8000, inclusive="both")]
bt
```
